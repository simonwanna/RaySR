```python
import numpy as np
import matplotlib.pyplot as plt
import sionna.rt as rt
```

```python
# Load a Sionna scene, inspect its bounding box, and add a transmitter for testing.
# If you have Sionna installed, run this cell to verify the scene loads.

# Load the default (empty) scene or change to a built-in one like rt.scene.etoile
scene = rt.load_scene(rt.scene.etoile)
print("Loaded scene:", scene)

# The underlying scene exposes a `mi_scene.bbox()` method (see MyScene wrapper in the repo).
try:
    bbox = scene.mi_scene.bbox()
    print("Scene bbox:")
    print(f"  x: {bbox.min.x} .. {bbox.max.x}")
    print(f"  y: {bbox.min.y} .. {bbox.max.y}")
    print(f"  z: {bbox.min.z} .. {bbox.max.z}")
except Exception as e:
    print("Could not read bbox from scene.mi_scene:", e)

# Set an operating frequency and add a simple transmitter to the scene
scene.frequency = 2.4e9
try:
    tx = rt.Transmitter('tx1', position=[0.0, 0.0, 10.0])
    scene.add(tx)
    print('Added transmitter tx1 at', tx.position)
except Exception as e:
    print('Could not add transmitter (class or API may differ):', e)

# Quick example (commented out) showing how to compute a small radio map with the RadioMapSolver.
# Note: the RadioMapSolver can be slow; keep samples_per_tx small for quick tests.
#
# solver = rt.RadioMapSolver()
# rm = solver(scene,
#             max_depth=3,
#             samples_per_tx=10000,
#             cell_size=(5.0, 5.0),
#             center=(0.0, 0.0, 0.0),
#             size=(100.0, 100.0))
# import matplotlib.pyplot as plt
# plt.imshow(np.max(rm.path_gain, axis=0), origin='lower', cmap='viridis')
# plt.colorbar(label='Path gain (dB)')
# plt.show()
```
